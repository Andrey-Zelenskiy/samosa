```python
import toml
import warnings
import pandas as pd

def custom_formatwarning(msg, *args, **kwargs):
    # ignore everything except the message
    return str(msg) + '\n'

warnings.formatwarning = custom_formatwarning
```

## Tools for lattice generation

__Desired properties:__

* Adjacency methods for nearest-neighbours along the principle lattice directions
* Vertex point group
* Bond point group

```python
lattice_lib = toml.load('../../include/lattices.toml')
```

```python
point_group_lib = toml.load('../../include/point_groups.toml')
```

```python
# Create a library that connects lattice point groups to lattices and space 
# groups

point_group_names = point_group_lib.keys()
lib_entries = [{'generators':[],
                'lattices':[],
                'space_groups':[]} for i in range(len(point_group_names))]
                        
crystal_symmetry_lib = dict(zip(point_group_names,lib_entries))

## Assign the point group generators
for pg in point_group_names:
    crystal_symmetry_lib[pg]['generators'] = point_group_lib[pg]['generators']

## Assign possible bravais lattices and space groups
lattice_names = lattice_lib.keys()

for l in lattice_names:
    allowed_pg = list(lattice_lib[l].keys())[1:]
    for pg in allowed_pg:
        crystal_symmetry_lib[pg]['lattices'] += [l]
        crystal_symmetry_lib[pg]['space_groups'] += lattice_lib[l][pg]['space_groups']
```

```python
# Create a library that provides lattice type and point group symmetry from the
# space group number

space_group_names = np.arange(-17,230)
space_group_names[17:] += 1

lib_entries = [{'lattice':'',
                'point_group':''} for i in range(len(space_group_names))]

space_group_lib = dict(zip(space_group_names,lib_entries))

for l in lattice_names:
    allowed_pg = list(lattice_lib[l].keys())[1:]
    for pg in allowed_pg:
        allowed_sg = lattice_lib[l][pg]['space_groups']
        for sg in allowed_sg:
            space_group_lib[sg]['lattice'] = l
            space_group_lib[sg]['point_group'] = pg
```

```python
class lattice:
    """
    Defines the properties of a periodic system. 
    The properies of a lattice object are

    n_sites         - int, total number of sites in the system;
    bravais_lattice - str, name of the Bravais lattice;
    basis_size      - int, number of basis vectors or the dimensionality of the 
                           lattice. Supported 1-3 for conventional lattices;
    l_array         - list or np.ndarray, for basis_size != 1, the dimensions
                                          of the lattice along the principle 
                                          directions; 
    
    """

    bravais_lattice = None
    point_group = None
    
    def __init__(self, dimensions, bravais_lattice = None, point_group = None):

        ## Assign lattice dimensions 
        dimensions_type = type(dimensions)
        
        if dimensions_type == np.ndarray or dimensions_type == list:
            dimensions_l = len(dimensions)
            
            self.n_sites = np.prod(dimensions)
            self.l_array = dimensions

        # If given a single integer, assign the total number of sites, raise
        # error for 2- and 3-dimensional lattices
        elif dimensions_type == int:
            if self.basis_size != 1:    
                raise TypeError('dimensions must be a list or np.ndarray '\
                              + 'of size {}'.format(self.basis_size)\
                              + ', not int')
                
            self.n_sites = dimensions

        else:
            raise TypeError('dimensions must be of type int, list, '\
                          + 'or np.ndarray, not ' + dimensions_type.__name__)

        ## Specify bravais lattice type, point group symmetry, and
        ## dimensionality of the lattice
        if bravais_lattice != None:
            self.set_lattice_type(bravais_lattice)
    
            ## Specify the lattice point group
            if point_group != None:
                self.set_point_group(point_group)

            else:
                point_group_list = lattice_lib[self.bravais_lattice].keys()
                point_group_list = list(point_group_list)[1:]
                supported_pg_names = ", ".join(point_group_list)
                
                warnings.warn('Point group not selected!'\
                            + '\nAllowed point groups are: ' + supported_pg_names)

        else:
            ## Specify the lattice point group
            if point_group != None:
                self.set_point_group(point_group)

                lattice_list = crystal_symmetry_lib[point_group]['lattices']
                supported_names = ", ".join(lattice_list)
                
                warnings.warn('Lattice type not selected!'\
                              '\nAllowed Bravais lattices are: '\
                            + supported_names + \
                              '\nSetting dimensionality to the size of'
                              ' the dimensions array')

            else:
                point_group_list = crystal_symmetry_lib.keys()
                point_group_list = list(point_group_list)
                supported_pg_names = ", ".join(point_group_list)

                warnings.warn('Lattice type not selected!'\
                              '\nSetting dimensionality to the size of'
                              ' the dimensions array')
                warnings.warn('Point group not selected!')

            self.basis_size = len(dimensions)


    def set_lattice_type(self, bravais_lattice):
        """
        Assigns lattice properties depending on the name of the Bravais lattice

        Arguments:

        bravais_lattice - str, name of the Bravais lattice.
        """

        if self.TestPointGroupSet():  
            lattice_list = crystal_symmetry_lib[self.point_group]['lattices']
            
            if bravais_lattice not in lattice_list:
                supported_names = ", ".join(lattice_list)
                raise NameError('Bravais lattice name must be one of: '\
                              + lattice_names)
        
        # Test if the new lattice properties match up with the existing ones
        bravais_lattice_new = bravais_lattice
        self.TestLatticeType(bravais_lattice_new)
        
        basis_size_new = lattice_lib[bravais_lattice]['dimension']
        self.TestLatticeDimensions(basis_size_new)

        # Record the new lattice name dimensionality
        self.bravais_lattice = bravais_lattice_new
        self.basis_size = basis_size_new

    def set_point_group(self, point_group):
        """
        Assigns point group based on the Bravais lattice type

        Arguments:

        point_group - str, name of the point group.
        """

        if self.TestLatticeSet():
            point_group_list = lattice_lib[self.bravais_lattice].keys()
            point_group_list = list(point_group_list)[1:]
            
            if point_group not in point_group_list:
                supported_pg_names = ", ".join(point_group_list)
                raise NameError('Point group must be one of: '\
                              + supported_pg_names)

        # Record the new lattice point group
        self.point_group = point_group

    
    def TestLatticeSet(self):
        """
        Check that the lattice type is selected
        """
        if self.bravais_lattice == None:
            return False
        else:
            return True

    def TestPointGroupSet(self):
        """
        Check that the point group is selected
        """
        if self.point_group == None:
            return False
        else:
            return True
    
    def TestLatticeType(self,bravais_lattice):
        """
        Check that the lattice type is one of the implemented lattices
        """
        if bravais_lattice not in lattice_lib.keys():
            supported_names = ", ".join(lattice_lib.keys())
            raise NameError('Incorrect lattice name: ' + bravais_lattice\
                          + '\nSupported lattices: ' + supported_names)
    
    def TestLatticeDimensions(self,basis_size):
        """
        Check that the l_array is of the right size and matches 
        the total number of sites
        """
        if len(self.l_array) != basis_size:
            raise TypeError('l_array has dimension '\
                          + '{}'.format(len(self.l_array))\
                          + ' but must be of dimension '\
                          + '{}'.format(basis_size))

        if np.prod(self.l_array) != self.n_sites:
            raise ValueError('Lattice dimensions in l_array do not sum to '\
                           + 'n_sites!')
    
    
    def __str__(self):

        base_string = 'Lattice object with {} sites'.format(self.n_sites)

        base_string += '\nDimension: {}'.format(self.basis_size)

        if self.basis_size == 2:
            base_string += '\nL1 = {}'.format(self.l_array[0])
            base_string += ', L2 = {}'.format(self.l_array[1])
        
        elif self.basis_size == 3:
            base_string += '\nL1 = {}'.format(self.l_array[0])
            base_string += ', L2 = {}'.format(self.l_array[1])
            base_string += ', L3 = {}'.format(self.l_array[2])
        
        if self.bravais_lattice != None:
            base_string += '\nBravais_lattice type: ' + self.bravais_lattice

        if self.point_group != None:
            base_string += '\nCrystal point group: ' + self.point_group
            
        return base_string
```

```python
new_lattice = lattice([2,3],bravais_lattice='hexagonal_p',point_group='C3')

print(new_lattice)
```
