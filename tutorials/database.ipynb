# General-purpose database  <a name="general-database"></a>

To simplify the manipulation of input data in samsa, `Database` class defines most necessary methods for analysing and loading available data files. 
Currently, the module is designed to read data from files with _.toml_ extension.

```python
"""
To have an overview of available files, initialize an empty Database object
"""

samsa_database = Database()
print(samsa_database)
```

```python
# View the absolute path to the samsa/database directory
print(samsa_database.path_to_database)
```

```python
"""
When initialized, Database creates a list of data files in the samsa/database
directory. All of these files can be uploaded to the database
"""

samsa_database.available_files
```

```python
"""
We can load the data files using either their names of integer index in the
order that they appear in Database.available_files
"""

# Load a file by specifying its name
samsa_database.load_file("point_groups.toml")

# Load a file by specifying index of the file in Database.availble_files
samsa_database.load_file(0)

print(samsa_database)
```

```python
"""
Once data is loaded, we can view it by specifying the name of the file
"""

samsa_database.data["point_groups.toml"]
```

# Space-group database  <a name="space-group-database"></a>

`SpaceGroupDatabase` is a sub-class od Database, which automatically uploads
files that contain data related to space-group properties.
On top of the simple methods of `Database`, this class defines functions to aid
space group analysis, such as tools for matching space groups, crystallographic
point groups, and lattice types.

__Note__: space group labels are given as numbers according to the
notation of the International Tables for Crystallography.

## Available data tables <a name="space-group-database-dictionaries"></a>

```python
"""
When a SpaceGroupDatabase is initialized, it automatically uploads the
necessary data files.
"""

sg_database = SpaceGroupDatabase()
print(sg_database)
```

```python
"""
The first file (space_group_lattices.toml) contains relationships between
dimensionality, lattice type, point group symmetry, and space group symmetry.

From this data, `SpaceGroupDatabase` creates lists of possible symmetry
properties, as well as compatibility between them.
"""

# All possible lattice types
print(f"Allowed lattice types:\n{', '.join(sg_database.lattice_list)}")

print("\n")

# All crystallographic Schoenflies symbols
print(f"Schoenflies symbols of 32 crystallographic point groups:\n\
{', '.join(sg_database.point_group_list)}")
```

```python
"""
Compatibility dictionaries end with '_reference'
"""

# Possible symmetry properties for a given lattice type
sg_database.lattice_reference['Hexagonal_P']
```

```python
# Possible symmetry properties for a given point group
sg_database.point_group_reference["D6"]
```

```python
# Possible symmetry properties for a given dimension and space group index
sg_database.space_group_reference[(3, 144)]
```

```python
# Possible symmetry properties for a given dimension
sg_database.dimension_reference[2]
```

```python
"""
In space_group_generators.toml, we store the generators of all space groups
"""

sg_database.space_group_generators[(3, 58)]
```

## Research tools <a name="space-group-database-tools"></a>

```python
"""
SpaceGroupDatabase provides a tool for manipulating the reference dictionaries
to identify a space group with desired symmetry properties. Depending on the
number of constrains provided, the search will return allowed combinations of
dimension/lattice-type/point-group.
"""

print("Constraint on lattice:")
print(sg_database.find_space_group(lattice_type="Hexagonal_P", as_str=True))

print("\nConstraint on point group and dimension:")
print(sg_database.find_space_group(dimension=3, point_group_symbol="C6", as_str=True))

print("\nConstraint on point group:")
print(sg_database.find_space_group(point_group_symbol="C6", as_str=True))

print("\nConstraint on lattice and point group:")
print(sg_database.find_space_group(lattice_type="Hexagonal_P",
                                   point_group_symbol="C6",
                                   as_str=True))
```

```python
# Search with no constraints on lattice type or point group produces a warning
print(sg_database.find_space_group(dimension=2, as_str=True))
```
