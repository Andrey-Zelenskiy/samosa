# Lattice structure  <a name="lattice"></a>

Implementations of point and space groups in `samosa` make it a perfect tool to
analyse lattice models.
The `Lattice` class in `samosa/structures` is designed to quickly initialize 
properties of a crystallographic lattice.

# Initialization methods  <a name="initialization"></a>

## From space group index <a name="initialization-space-group"></a>

```python
"""
The standard way to initialize a lattice is to provide the dimension of the
lattice and space group number
"""

lattice = Lattice(dimension=2, space_group_index=15)
print(lattice)
```

```python
"""
By default, the vertices of the lattice are assumed to be at [0, 0, 0] position
(+ equivalent symmetrical sites). We can choose to place the vertices
differently inside the unit cell by providing a set of fractional coordinates
"""

lattice = Lattice(dimension=2, 
                  space_group_index=15,
                  positions=[[0, 0, 0], [0.5, 0, 0]])
print(lattice)
```

```python
print(f"Inequivalent Wyckoff sites are\n\n{lattice.wyckoff_list[0]},\
      \n\n{lattice.wyckoff_list[1]}")
```

```python
"""
It is also possible to set the unconstrained crystal parameters 
(unit cell dimensions and angles).
`random_parameters` attribute reflects if the unit cell parameters were
initialized at random or from user input.
"""

lattice = Lattice(dimension=2, 
                  space_group_index=15, 
                  crystal_parameters={"a": 1.3})
print(lattice)

# lattice.random_parameters # will output False
```

```python
"""
Once the lattice is initialized, we can check what the unconstrained parameters
are and set them accordingly.
"""

lattice = Lattice(dimension=3, space_group_index=15)

# Uncomment to print the initial parameters of the lattice object
#print(f"{lattice}\n\n")

print(f"The constrained parameters are\
      \n{lattice.constrained_parameters}\n")
print(f"The unconstrained parameters are\
      \n{lattice.unconstrained_parameters}")

lattice.set_crystal_parameters({"a": 1.0, "b": 0.8, "c": 1.5, "beta": 42})

print(f"\n\n{lattice}")
```

## Simple lattices <a name="initialization-simple"></a>

Often, one is interested in constructing a simple lattice (for example square),
and the exact space group is not known, or is tedious to determine. 
`samosa` provides initialization of simple lattices: (1D) chain, (2D) square,
hexagonal, (3D) cubic, bcc (base-centered cubic), and fcc 
(face-centered cubic).

```python
"""
To initialize a simple lattice, we still need to input a point group, since the
point group symmetry of the lattice can be composed of either only rotations,
or rotations and reflections.
"""

cubic_lattice = Lattice.initialize_simple("cubic", point_group="O")
print(cubic_lattice)
```
